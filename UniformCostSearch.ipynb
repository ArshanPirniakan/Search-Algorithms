```python
# Uniform Cost Search (UCS) Implementation
# Finds the path with the lowest cumulative cost

# Graph represented as an adjacency list with (neighbor, edge_cost)
graph = {
    'C': [('B', 5), ('T', 4), ('O', 6), ('E', 2), ('P', 8)],
    'B': [('A', 3), ('R', 5), ('S', 1)],
    'T': [],
    'O': [('I', 9), ('N', 7)],
    'E': [('G', 4)],
    'P': [('L', 2), ('F', 2), ('D', 3)],
    'A': [],
    'R': [],
    'S': [],
    'I': [('Z', 1)],
    'N': [],
    'G': [],
    'L': [],
    'F': [],
    'D': [],
    'Z': []
}

def ucs(graph, root, Goals):
    # Dictionary to keep track of visited nodes and their minimum path cost
    ClosedDict = {} 
    # Priority Queue: (node, parent, cumulative_cost)
    Queue = []   
    
    # Initialize Queue with the root node (cost = 0)
    Queue.append((root, None, 0)) 
  
    print("Nodes expanded:", end=" ")
    while Queue:
        # Pop the node with the lowest cumulative path cost (front of sorted list)
        s = Queue.pop(0)         
        print(s[0], end=" ")
        
        # GOAL CHECK
        if s[0] in Goals:
            print(f"\n\nA path to goal state '{s[0]}' is found with path cost: {s[2]}")
            
            # Reconstruct the path using ClosedDict
            Path = []
            curr = s
            Path.append(curr)
            while curr[0] != root:
                # Retrieve parent information
                parent_data = ClosedDict[curr[1]]
                curr = (curr[1], parent_data[0], parent_data[1])
                Path.insert(0, curr)
            
            # Print the formatted path
            for i, node in enumerate(Path):
                if i == len(Path) - 1:
                    print(f"({node[1]}, {node[0]}, {node[2]})")
                else:
                    print(f"({node[1]}, {node[0]}, {node[2]}) -> ", end='')          
            break
          
        # Expand neighbors
        for neighbour, edge_cost in graph[s[0]]:
            new_cost = s[2] + edge_cost
            new_node = (neighbour, s[0], new_cost)
            
            # Insert neighbor into Queue in sorted order of path costs
            inserted = False
            for j in range(len(Queue)):
                if new_node[2] <= Queue[j][2]:
                    Queue.insert(j, new_node)
                    inserted = True
                    break
            if not inserted:
                Queue.append(new_node)
        
        # Mark node as closed with its parent and cost
        ClosedDict.update({s[0]: (s[1], s[2])})

# Run the UCS algorithm
ucs(graph, 'C', Goals=('L', 'Z'))
```
