```python
# Best-First Search Implementation
# Heuristic-based search algorithm

# Define the graph as an adjacency list
graph = {
    'C': ['B', 'T', 'O', 'E', 'P'],
    'B': ['A', 'R', 'S'],
    'T': [],
    'O': ['I', 'N'],
    'E': ['G'],
    'P': ['L', 'F', 'D'],
    'A': [],
    'R': [],
    'S': [],
    'I': ['Z'],
    'N': [],
    'G': [],
    'L': [],
    'F': [],
    'D': [],
    'Z': []
}

# Heuristic values (estimated distance to goal)
h_func = {
    'A': 17, 'B': 14, 'C': 21, 'D': 100, 'E': 12, 'F': 8, 
    'G': 51, 'I': 4, 'L': 0, 'N': 44, 'O': 7, 'P': 15, 'R': 20,
    'S': 50, 'T': 5, 'Z': 0
}

def best_fs(graph, h_func, root, Goals):
    ClosedDict = {}  # Keeps track of visited nodes and their parents
    Queue = []       # Priority queue (list of (node, parent, h-value))
    
    # Initialize queue with the root node
    Queue.append((root, None, h_func[root])) 
  
    print("Nodes expanded:", end=" ")
    while Queue:
        # Pop the node with the lowest heuristic value (front of sorted list)
        s = Queue.pop(0)
        print(s[0], end=" ")

        # GOAL CHECK
        if s[0] in Goals:
            print(f"\n\nA path to goal state '{s[0]}' was found:")
            
            # Reconstruct the path using the ClosedDict
            Path = [s]
            curr = s
            while curr[0] != root:
                parent_info = ClosedDict[curr[1]]
                curr = (curr[1], parent_info[0], parent_info[1])
                Path.insert(0, curr)
            
            # Print the formatted path
            for i, node in enumerate(Path):
                connector = " -> " if i < len(Path) - 1 else ""
                print(f"({node[1]}, {node[0]}, {node[2]})", end=connector)
            break
      
        # Expand neighbors
        for neighbour in graph[s[0]]:
            # Check if neighbor is already visited or in queue
            in_closed = neighbour in ClosedDict
            in_queue = any(node[0] == neighbour for node in Queue)
                    
            if not in_closed and not in_queue:
                # Insert into Queue maintaining sorted order of h-values
                new_entry = (neighbour, s[0], h_func[neighbour])
                inserted = False
                for i in range(len(Queue)):
                    if h_func[neighbour] <= Queue[i][2]:
                        Queue.insert(i, new_entry)
                        inserted = True
                        break
                if not inserted:
                    Queue.append(new_entry)

        # Mark current node as closed
        ClosedDict.update({s[0]: (s[1], h_func[s[0]])})

# Run the algorithm
best_fs(graph, h_func, 'C', Goals=('L', 'Z'))
```
