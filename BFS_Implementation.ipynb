```python
# Breadth-First Search (BFS) Implementation
# Explores the graph level-by-level (FIFO logic)

# Define the graph as an adjacency list
graph = {
    'C': ['B', 'T', 'O', 'E', 'P'],
    'B': ['A', 'R', 'S'],
    'T': [],
    'O': ['I', 'N'],
    'E': ['G'],
    'P': ['L', 'F', 'D'],
    'A': [],
    'R': [],
    'S': [],
    'I': ['Z'],
    'N': [],
    'G': [],
    'L': [],
    'F': [],
    'D': [],
    'Z': []
}

def breadth_fs(graph, root, Goals):
    # Dictionary to keep track of visited nodes and their parents
    ClosedDict = {} 
    # Queue for First-In-First-Out (FIFO) node exploration
    Queue = []   
    
    # Initialize Queue with (current_node, parent)
    Queue.append((root, None)) 
  
    print("Expansion Order:", end=" ")
    while Queue:
        # Pop the first element added to the queue (FIFO)
        s = Queue.pop(0)      
        print(s[0], end=" ")

        # GOAL CHECK: Check if the current node is one of our targets
        if s[0] in Goals:
            print(f"\n\nA path to goal state '{s[0]}' is found:")
            
            # Backtrack from goal to root to reconstruct the path
            Path = [s]
            curr = s
            while curr[0] != root:
                parent = ClosedDict[curr[1]]
                curr = (curr[1], parent)
                Path.insert(0, curr)
            
            # Print the path clearly
            for i, node in enumerate(Path):
                if i == len(Path) - 1:
                    print(f"({node[1]}, {node[0]})")
                else:
                    print(f"({node[1]}, {node[0]}) -> ", end='')          
            break
      
        # Check all neighbors of the current node
        for neighbour in graph[s[0]]:
            # Check if neighbor has already been visited or is already in the queue
            is_closed = neighbour in ClosedDict
            is_in_queue = any(node[0] == neighbour for node in Queue)
                    
            if not is_closed and not is_in_queue:
                Queue.append((neighbour, s[0]))
        
        # Mark the current node as closed by storing its parent
        ClosedDict.update({s[0]: s[1]})

# Run the BFS algorithm starting at 'C' searching for 'L' or 'Z'
breadth_fs(graph, 'C', Goals=('L', 'Z'))
```
