```python
# Depth-First Search (DFS) Implementation
# Explores the graph by going as deep as possible before backtracking

# Graph represented as an adjacency list
graph = {
    'C': ['B', 'T', 'O', 'E', 'P'],
    'B': ['A', 'R', 'S'],
    'T': [],
    'O': ['I', 'N'],
    'E': ['G'],
    'P': ['L', 'F', 'D'],
    'A': [],
    'R': [],
    'S': [],
    'I': ['Z'],
    'N': [],
    'G': [],
    'L': [],
    'F': [],
    'D': [],
    'Z': []
}

def dfs_visit(graph, ClosedDict, Queue, Goals, VisitedNodes):    
    # Pop the first node (the most recently added neighbor)
    node = Queue.pop(0)
    print(node[0], end=' ')
    
    # Check if the current node is a goal
    if node[0] in Goals:
        ClosedDict.update({node[0]: node[1]})
        # Insert a marker to stop further exploration down this branch
        Queue.insert(0, (None, None))
    else:
        for neighbour in graph[node[0]]: 
            CheckFlag = False
            if neighbour in ClosedDict:
                CheckFlag = True
            else:
                for q_node in Queue:
                    if q_node[0] == neighbour:
                        CheckFlag = True
            
            # If not visited, recurse deeper immediately (LIFO behavior)
            if not CheckFlag:
                if len(Queue) == 0 or Queue[0] != (None, None):
                    Queue.insert(0, (neighbour, node[0]))
                    VisitedNodes.update({neighbour: node[0]})
                    dfs_visit(graph, ClosedDict, Queue, Goals, VisitedNodes)
                else:
                    # Stop expanding neighbors if a goal was found in a deeper branch
                    break
        
        ClosedDict.update({node[0]: node[1]}) 
    return VisitedNodes

def dfs(graph, root, Goals):
    ClosedDict = {}     # Nodes that have been fully processed
    Queue = [(root, None)] # Stack-like queue for tracking next nodes
    VisitedNodes = {root: None}
    
    print("Expansion Order:", end=" ")
    return dfs_visit(graph, ClosedDict, Queue, Goals, VisitedNodes)

# Execution
Root = 'C'
Goals = ['L', 'Z']          
VisitedNodes = dfs(graph, Root, Goals)

# Path Reconstruction and Visualization
print("\n" + "-"*30)
for Goal in Goals:
    if Goal in VisitedNodes:
        Path = []
        curr_node = (Goal, VisitedNodes[Goal])
        Path.append(curr_node)
        
        # Trace back to root
        while curr_node[0] != Root:            
            parent = VisitedNodes[curr_node[1]]
            curr_node = (curr_node[1], parent)
            Path.insert(0, curr_node)            
        
        print(f"Path to goal '{Goal}':")
        for i, node in enumerate(Path):
            if i == len(Path) - 1:
                print(f"({node[1]}, {node[0]})")
            else:
                print(f"({node[1]}, {node[0]}) -> ", end='')
```
