```python
# Iterative Deepening Depth-First Search (IDDFS) Implementation
# Combines DFS (space efficiency) with BFS (optimality)

# Graph represented as an adjacency list
graph = {
    'C': ['B', 'T', 'O', 'E', 'P'],
    'B': ['A', 'R', 'S'],
    'T': [],
    'O': ['I', 'N'],
    'E': ['G'],
    'P': ['L', 'F', 'D'],
    'A': [],
    'R': [],
    'S': [],
    'I': ['Z'],
    'N': [],
    'G': [],
    'L': [],
    'F': [],
    'D': [],
    'Z': []
}

def dfs_visit(graph, ClosedDict, Queue, Goals, Depth_Limit):    
    # Pop the last added neighbor (LIFO behavior)
    node = Queue.pop(0)
    print(node[0], end=' ')
    
    # Goal check
    if node[0] in Goals:
        ClosedDict.update({node[0]: (node[1], node[2])})
        ClosedDict.update({True: '_'}) # Marker for found goal
        Queue.insert(0, (None, None, None))
    else:
        for neighbour in graph[node[0]]: 
            # Cycle detection
            CheckFlag = False
            if neighbour in ClosedDict:
                CheckFlag = True
            else:
                for Record in Queue:
                    if Record[0] == neighbour and Record[1] == node[0]:
                        CheckFlag = True
            
            # If not visited, check depth limit and proceed deeper
            if not CheckFlag:
                current_depth = node[2] + 1
                if current_depth <= Depth_Limit:
                    if len(Queue) == 0 or Queue[0] != (None, None, None):
                        Queue.insert(0, (neighbour, node[0], current_depth))
                        dfs_visit(graph, ClosedDict, Queue, Goals, Depth_Limit)
                    else:
                        break
        
        ClosedDict.update({node[0]: (node[1], node[2])}) 
    return ClosedDict

def dfs(graph, root, Goals, Depth_Limit):
    ClosedDict = {}       # Track completed nodes and their parents
    Queue = [(root, None, 0)] # Stack for LIFO exploration (node, parent, depth)
    return dfs_visit(graph, ClosedDict, Queue, Goals, Depth_Limit)

# --- Main Execution ---
Root = 'C'
Goals = ['L']
Max_Depth = 3

print("Starting Iterative Deepening Search...")

# Iteratively increase depth limit
for Depth in range(Max_Depth + 1):
    print(f"\nSearching at Depth Limit: {Depth}")
    ClosedDict = dfs(graph, Root, Goals, Depth)
    
    # Check if a goal was found in this iteration
    if True in ClosedDict:
        print(f"\nGoal '{Goals[0]}' found at depth {Depth}!")
        
        # Path Reconstruction
        Path = []
        curr_node = Goals[0]
        while curr_node != Root:
            parent_info = ClosedDict[curr_node]
            Path.insert(0, (parent_info[0], curr_node, parent_info[1]))
            curr_node = parent_info[0]
        
        # Print path
        print("Path: ", end="")
        for i, node in enumerate(Path):
            if i == len(Path) - 1:
                print(f"({node[0]}, {node[1]}, {node[2]})")
            else:
                print(f"({node[0]}, {node[1]}, {node[2]}) -> ", end='')
        break
```
